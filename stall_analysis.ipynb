```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
worst_day = days[worst_index]
best_week = best_index // 7 + 1
worst_week = worst_index // 7 + 1

print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest daily sales: KES {highest_sales:,.2f} ({best_day}, Week {best_week})")
print(f"Lowest daily sales: KES {lowest_sales:,.2f} ({worst_day}, Week {worst_week})")
```

```python
def week_totals(sales):
    """Return the total sales for week 1 and week 2 as a tuple."""
    week_1_total = sum(sales[:7])
    week_2_total = sum(sales[7:14])
    return week_1_total, week_2_total


week_1, week_2 = week_totals(daily_sales)
week_on_week_growth = ((week_2 - week_1) / week_1) * 100

print(f"Week 1 sales: KES {week_1:,.2f}")
print(f"Week 2 sales: KES {week_2:,.2f}")
print(f"Week-on-week growth: {week_on_week_growth:.2f}%")
```

```python
def classify_day(amount, average):
    """Classify sales as Strong, Normal, or Weak relative to the average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


category_counts = {"Strong": 0, "Normal": 0, "Weak": 0}
daily_categories = []

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_categories.append(category)
    category_counts[category] += 1

for category, count in category_counts.items():
    print(f"{category} days: {count}")
```

```python
weekday_sales = []
weekend_sales = []

for index, amount in enumerate(daily_sales):
    if days[index] in ["Sat", "Sun"]:
        weekend_sales.append(amount)
    else:
        weekday_sales.append(amount)

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

weekend_difference_percent = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print(f"Average weekday sales: KES {weekday_average:,.2f}")
print(f"Average weekend sales: KES {weekend_average:,.2f}")
print(f"Weekend average compared with weekday average: {weekend_difference_percent:.2f}% higher")
```

```python
def products_to_reorder(stock_items):
    """Return product names with stock at or below their reorder level."""
    reorder_list = []

    for product, details in stock_items.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


reorder_list = products_to_reorder(stock)

print("Products to reorder:")
for product in reorder_list:
    print(f"- {product}: {stock[product]['qty']} kg in stock "
          f"(reorder level: {stock[product]['reorder']} kg)")

print("\\nGross margin by product:")
for product, details in stock.items():
    gross_margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100
    print(f"{product}: {gross_margin:.2f}%")
```

```python
print("=" * 52)
print("           STALL PERFORMANCE REPORT")
print("             Wanjiku — Gikomba Market")
print("=" * 52)

print(f"Period:                 14 days")
print(f"Total sales:            KES {total_sales:,.2f}")
print(f"Average daily sales:    KES {average_sales:,.2f}")
print(f"Highest sales:          KES {highest_sales:,.2f} "
      f"({best_day}, Week {best_week})")
print(f"Lowest sales:           KES {lowest_sales:,.2f} "
      f"({worst_day}, Week {worst_week})")

print("-" * 52)
print(f"Week 1 total:           KES {week_1:,.2f}")
print(f"Week 2 total:           KES {week_2:,.2f}")
print(f"Week-on-week growth:    {week_on_week_growth:.2f}%")

print("-" * 52)
print(f"Weekday average:        KES {weekday_average:,.2f}")
print(f"Weekend average:        KES {weekend_average:,.2f}")
print(f"Weekend uplift:         {weekend_difference_percent:.2f}%")

print("-" * 52)
print("DAILY PERFORMANCE COUNTS")
for category, count in category_counts.items():
    print(f"{category:<10}: {count} days")

print("-" * 52)
print("RESTOCKING PRIORITIES")
for product in reorder_list:
    details = stock[product]
    print(f"{product}: {details['qty']} kg available; "
          f"reorder level {details['reorder']} kg")

print("-" * 52)
print("GROSS MARGINS")
for product, details in stock.items():
    margin = (details["price"] - details["cost"]) / details["price"] * 100
    print(f"{product:<10}: {margin:.2f}%")

print("=" * 52)
```
